```python
import pandas as pd
import threading
import queue
import requests
from ntscraper import Nitter

# Load proxies from valid proxies file
def load_proxies(proxy_file="valid_proxies.txt"):
    with open(proxy_file, "r") as f:
        return [proxy.strip() for proxy in f.readlines() if proxy.strip()]

# Get a proxy from the queue
def get_proxy(proxy_queue):
    try:
        return proxy_queue.get_nowait()
    except queue.Empty:
        return None

# Create a Nitter scraper with manual proxy handling
def create_scraper(proxy=None):
    scraper = Nitter()
    if proxy:
        scraper.proxies = {"http": proxy, "https": proxy}  # Manually assign proxy
    return scraper

# Define the three parties and their search criteria
parties = {
    'CCM': {
        'account': 'ccm_tanzania',
        'hashtags': ['#CCMTanzania', '#KaziIendelee', '#MamaAsemewe'],
        'leader': 'SamiaSuluhuHassan'
    },
    'CHADEMA': {
        'account': 'ChademaTz',
        'hashtags': ['#Chadema', '#PeoplesPower', '#ChademaMkutanoMkuu2025', '#LissuWillFixIt', '#NguvuYaUmma'],
        'leader': 'TunduLissu'
    },
    'ACT-Wazalendo': {
        'account': 'ACTwazalendo',
        'hashtags': ['#ACTWazalendo', '#TanzaniaForward'],
        'leader': 'ZittoKabwe'
    }
}

# Function to collect and save tweets
def collect_and_save_tweets(party_name, search_data, proxy_queue, num_tweets=300):
    all_tweets = []
    proxy = get_proxy(proxy_queue)
    scraper = create_scraper(proxy)

    print(f"🔍 Using proxy: {proxy} for {party_name}")

    # Collect from official account
    try:
        all_tweets += scraper.get_tweets(search_data['account'], mode='user', number=num_tweets // 3)
    except Exception as e:
        print(f"⚠️ Error collecting tweets from account: {search_data['account']} - {e}")

    # Collect from hashtags
    for hashtag in search_data['hashtags']:
        try:
            all_tweets += scraper.get_tweets(hashtag, mode='hashtag', number=num_tweets // 3)
        except Exception as e:
            print(f"⚠️ Error collecting tweets for hashtag: {hashtag} - {e}")

    # Collect from leader mentions
    try:
        all_tweets += scraper.get_tweets(search_data['leader'], mode='mention', number=num_tweets // 3)
    except Exception as e:
        print(f"⚠️ Error collecting tweets mentioning: {search_data['leader']} - {e}")

    # Save tweets to CSV
    if all_tweets:
        df = pd.DataFrame(all_tweets)
        file_path = f"data/{party_name}_tweets.csv"
        df.to_csv(file_path, index=False)
        print(f"✅ Collected {len(df)} tweets for {party_name} and saved to {file_path}")
    else:
        print(f"❌ No tweets collected for {party_name}")

# Main function to start scraping
def main():
    # Load valid proxies
    proxy_list = load_proxies("valid_proxies.txt")
    proxy_queue = queue.Queue()

    for proxy in proxy_list:
        proxy_queue.put(proxy)

    # Start scraping with threads for each party
    threads = []
    for party, info in parties.items():
        t = threading.Thread(target=collect_and_save_tweets, args=(party.replace("-", ""), info, proxy_queue))
        t.start()
        threads.append(t)

    # Wait for all threads to finish
    for t in threads:
        t.join()

    print("🚀 All parties' tweets collected successfully!")

if __name__ == "__main__":
    main()
```
